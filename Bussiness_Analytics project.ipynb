```python
#Exploratory data analysis(EDA)

df.describe()
```

```python
#sales distribution
plt.hist(df["Sales"])
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution")
plt.show()
```

```python
#profit Distribution

plt.hist(df["Profit"])
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.title("Sales Distribution")
plt.show()
```

```python
#identity bussiness variable
df[["Quantity","Sales","Profit"]].head().round()
```

```python
#pearson's Correlation Analysis

pearson_corr=df[["Quantity","Sales","Profit"]].corr(method="pearson")
print(pearson_corr)
```

```python
#spearman rank correlation analysis

spearman_corr=df[["Quantity","Sales","Profit"]].corr(method="spearman")
print(spearman_corr)
```

```python
#scatter plot
plt.scatter(df["Quantity"],df["Sales"])
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.title("Quantity VS Sales")
plt.show()
```

```python
#correlation  Heatmap

corr =df[["Quantity","Sales","Profit"]].corr()

sns.heatmap(corr,annot=True)
plt.title("correlation Heatmap")
plt.show()
```

```python
#Scatter plot matrix/pair plot

sns.pairplot(df[["Quantity","Sales","Profit"]])
plt.show()
```
